```python
# AgroVision: Check required Python libraries

import importlib.util

libraries = [
    "numpy",
    "pandas",
    "matplotlib",
    "PIL",
    "sklearn",
    "torch",
    "torchvision",
    "cv2",
    "fastapi",
    "jupyter"
]

print("AGROVISION - LIBRARY CHECK")
print("-" * 40)

for library in libraries:
    installed = importlib.util.find_spec(library) is not None

    status = "INSTALLED" if installed else "NOT INSTALLED"



    print(f"{library:15} : {status}")
```
